```r
library(DBI)
con <- dbConnect(RSQLite::SQLite(), "school.db")
dbWriteTable(con, "students", students, overwrite = TRUE)
dbGetQuery(con, "SELECT * FROM students")
dbGetQuery(con, "SELECT name FROM students WHERE marks > 80")
dbDisconnect(con)
```
